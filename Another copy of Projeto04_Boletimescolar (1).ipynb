```python

# ==========================================
#       BOLETIM ESCOLAR DIGITAL
# ==========================================

ARQUIVO = "boletim.txt"


# 1. CALCULAR MÉDIA
def calcular_media(notas):
    return sum(notas) / 3


# 2. DEFINIR SITUAÇÃO DO ALUNO
def situacao(media):
    if media >= 7:
        return "Aprovado"
    elif media >= 5:
        return "Recuperação"
    else:
        return "Reprovado"


# 3. LER UMA NOTA VÁLIDA
def ler_nota(numero):
    while True:
        try:
            nota = float(input(f"Nota {numero} (0 a 10): "))

            if 0 <= nota <= 10:
                return nota

            print("Nota inválida! Digite um valor entre 0 e 10.")

        except ValueError:
            print("Digite um número válido.")


# 4. SALVAR ALUNOS NO ARQUIVO
def salvar_alunos(alunos):
    with open(ARQUIVO, "w", encoding="utf-8") as arquivo:
        for aluno in alunos:
            linha = (
                f"{aluno['nome']};"
                f"{aluno['notas'][0]};"
                f"{aluno['notas'][1]};"
                f"{aluno['notas'][2]}\n"
            )
            arquivo.write(linha)


# 5. CARREGAR ALUNOS DO ARQUIVO
def carregar_alunos():
    alunos = []

    try:
        with open(ARQUIVO, "r", encoding="utf-8") as arquivo:
            for linha in arquivo:
                dados = linha.strip().split(";")

                if len(dados) != 4:
                    continue

                try:
                    nome = dados[0]
                    notas = [float(n) for n in dados[1:]]

                    if nome and ";" not in nome and all(0 <= n <= 10 for n in notas):
                        alunos.append({
                            "nome": nome,
                            "notas": notas
                        })

                except ValueError:
                    continue

    except FileNotFoundError:
        pass

    return alunos


# 6. BUSCAR ALUNO PELO NOME
def buscar_aluno(alunos, nome):
    for aluno in alunos:
        if aluno["nome"].lower() == nome.strip().lower():
            return aluno

    return None


# 7. MOSTRAR BOLETIM DE UM ALUNO
def mostrar_aluno(aluno):
    media = calcular_media(aluno["notas"])

    print(f"\nAluno: {aluno['nome']}")
    print(f"Notas: {aluno['notas']}")
    print(f"Média: {media:.2f}")
    print(f"Situação: {situacao(media)}")


# 8. CADASTRAR ALUNO
def cadastrar_aluno(alunos):
    nome = input("Nome do aluno: ").strip().title()

    if not nome or ";" in nome:
        print("Nome inválido!")
        return

    if buscar_aluno(alunos, nome) is not None:
        print("Esse aluno já está cadastrado!")
        return

    notas = []

    for i in range(1, 4):
        notas.append(ler_nota(i))

    aluno = {
        "nome": nome,
        "notas": notas
    }

    alunos.append(aluno)

    salvar_alunos(alunos)

    print("Aluno cadastrado e salvo!")


# 9. LISTAR TODOS OS ALUNOS
def listar_boletim(alunos):
    if not alunos:
        print("Nenhum aluno cadastrado.")
        return

    print("\n========== BOLETIM ESCOLAR ==========")

    for aluno in alunos:
        mostrar_aluno(aluno)


# 10. BUSCAR E MOSTRAR ALUNO
def menu_buscar(alunos):
    if not alunos:
        print("Nenhum aluno cadastrado.")
        return

    nome = input("Digite o nome do aluno: ")

    aluno = buscar_aluno(alunos, nome)

    if aluno:
        mostrar_aluno(aluno)
    else:
        print("Aluno não encontrado.")


# 11. ESTATÍSTICAS DA TURMA
def estatisticas(alunos):
    if not alunos:
        print("Nenhum aluno cadastrado.")
        return

    medias = [calcular_media(a["notas"]) for a in alunos]

    media_turma = sum(medias) / len(medias)

    maior_media = max(medias)

    melhor_aluno = alunos[medias.index(maior_media)]

    aprovados = 0
    recuperacao = 0
    reprovados = 0

    for media in medias:
        resultado = situacao(media)

        if resultado == "Aprovado":
            aprovados += 1
        elif resultado == "Recuperação":
            recuperacao += 1
        else:
            reprovados += 1

    print("\n========== ESTATÍSTICAS ==========")
    print(f"Média da turma: {media_turma:.2f}")
    print(f"Maior média: {melhor_aluno['nome']} ({maior_media:.2f})")
    print(f"Aprovados: {aprovados}")
    print(f"Recuperação: {recuperacao}")
    print(f"Reprovados: {reprovados}")


# 12. EXPORTAR RELATÓRIO
def exportar_relatorio(alunos):
    with open("relatorio.txt", "w", encoding="utf-8") as arquivo:

        for aluno in alunos:
            media = calcular_media(aluno["notas"])
            resultado = situacao(media)

            arquivo.write(
                f"{aluno['nome']};{media:.2f};{resultado}\n"
            )

    print("Relatório exportado para relatorio.txt!")


# ==========================================
#             PROGRAMA PRINCIPAL
# ==========================================

def main():

    alunos = carregar_alunos()

    while True:

        print("\n=== BOLETIM ESCOLAR DIGITAL ===")
        print("1 - Cadastrar aluno")
        print("2 - Listar boletim")
        print("3 - Buscar aluno")
        print("4 - Estatísticas da turma")
        print("5 - Exportar relatório")
        print("0 - Sair")

        opcao = input("Escolha: ")

        if opcao == "1":
            cadastrar_aluno(alunos)

        elif opcao == "2":
            listar_boletim(alunos)

        elif opcao == "3":
            menu_buscar(alunos)

        elif opcao == "4":
            estatisticas(alunos)

        elif opcao == "5":
            exportar_relatorio(alunos)

        elif opcao == "0":
            print("Programa encerrado!")
            break

        else:
            print("Opção inválida!")


main()

```
